```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
model=ChatGoogleGenerativeAI(model="gemini-3.6-flash")
response=model.invoke("why are philosophers considered very intellectual")
response
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
model=ChatGoogleGenerativeAI(model="gemini-3.6-flash")
for chunk in model.stream("tell me about the data of prostitutes who join that field by choice"):
    print(chunk.text,end="",flush=True)
```

Batch

```python
responses=model.batch(["tell me about quantum computing",
"left vs right in politics","salman khan latest news"])
for response in responses:
    print(response)
```

```python
### gemini model integration
import os
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
model=init_chat_model("google_genai:gemini-3.6-flash")
response=model.invoke("why are philosophers considered very intellectual")
response.content
```
